### Bước 4: Bắt đầu Huấn Luyện (Fine-tuning YOLOv8n)

```python
from ultralytics import YOLO

# Nạp mô hình pretrained nhẹ YOLOv8 Nano
model = YOLO('yolov8n.pt')

# Train 50 epochs (Thời gian khoảng 20-30 phút trên GPU T4)
results = model.train(
    data='data.yaml',   # Hoặc đường dẫn dataset.location + '/data.yaml'
    epochs=50,
    imgsz=640,
    batch=16,
    device=0,
    plots=True
)
```

### Bước 5: Đánh giá độ chính xác (mAP50-95)

```python
# Đánh giá mô hình trên tập test
metrics = model.val()
print(f"mAP50-95: {metrics.box.map}")
print(f"mAP50: {metrics.box.map50}")
```
